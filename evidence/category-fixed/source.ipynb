# Explicit group semantics preserve the report
Original Floe validation fixture.

```
import pandas as pd
orders = pd.read_csv('orders.csv')
orders['region'] = pd.Categorical(orders['region'], categories=['east', 'west', 'north'])
```

```
revenue = orders.groupby('region', observed=False)['revenue'].sum().reset_index()
print('Revenue report generated')
```
